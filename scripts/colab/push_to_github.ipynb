```
import getpass, subprocess
TOKEN = getpass.getpass('GitHub token: ')
remote = f'https://{GITHUB_USER}:{TOKEN}@github.com/{GITHUB_USER}/{REPO}.git'
cmds = [
 'git init -q', 'git branch -M main',
 'git config user.email "you@example.com"', 'git config user.name "you"',
 'git add .', 'git commit -q -m "Initial MVP"',
 f'git remote add origin {remote}', 'git push -u origin main',
]
for c in cmds:
    r = subprocess.run(c, shell=True, capture_output=True, text=True)
    print('$', c.replace(TOKEN, '***'), '\n', (r.stdout + r.stderr).replace(TOKEN, '***'))
```

Done. Open your repo on github.com -> **Actions** tab: the *build-apk* workflow starts automatically and fetches llama.cpp itself. When it finishes, open the run and download the **offline-research-debug-apk** artifact.
Delete the token on GitHub (Settings -> Developer settings) when you are finished.
